#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
mm_summary<- mm_tidy|>
summarize(year_mean = mean(year, na.rm = TRUE), #finds the mean, max, min for year
         year_max = max(year, na.rm = TRUE),
         year_min = min(year, na.rm = TRUE),
           seed_mean = mean(seed, na.rm = TRUE),#finds the mean, max, min for seed
    seed_min = min(seed, na.rm = TRUE),
    seed_max = max(seed, na.rm = TRUE),
    reg_wins_pct_mean = mean(reg_wins_pct, na.rm = TRUE),#finds the mean, max, min for regular season win percentage
    reg_wins_pct_min = min(reg_wins_pct, na.rm = TRUE),
    reg_wins_pct_max = max(reg_wins_pct, na.rm = TRUE),
    tourney_wins_mean = mean(tourney_wins, na.rm = TRUE),#finds the mean, max, min for tournament wins
    tourney_wins_min = min(tourney_wins, na.rm = TRUE),
    tourney_wins_max = max(tourney_wins, na.rm = TRUE))

mm_summary_tidy<- pivot_longer(mm_summary, 
             cols = year_mean:tourney_wins_max, # select all the columns that I want to reorganize
             names_to = "Variable", # new column that now holds the old column names
             values_to = "Value") #new column that holds the numbers
mm_summary_tidy
```
